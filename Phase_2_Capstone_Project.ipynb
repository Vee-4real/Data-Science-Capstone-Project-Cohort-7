### Verifying Raw Data Path Contents

Now that `RAW_DATA_path` is correctly defined, let's list the contents of this folder to confirm your CSV files are present.

```python
import os

try:
    print(f"Files in '{RAW_DATA_path}':")
    if os.path.exists(RAW_DATA_path):
        for item in os.listdir(RAW_DATA_path):
            print(repr(item))
    else:
        print(f"Error: The directory '{RAW_DATA_path}' was not found. Please ensure the 'DATA' folder exists inside '{PROJECT_FOLDER}' in your Google Drive.")
except NameError:
    print("Error: RAW_DATA_path is not defined. Please ensure the cell defining PROJECT_FOLDER and RAW_DATA_path has been run.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

```python
# Load the CSV files
# Make sure RAW_DATA_path is defined and points to the correct folder containing your CSV files.
try:
    customers = pd.read_csv(
        os.path.join(RAW_DATA_path, 'customers.csv')
    )

    products = pd.read_csv(
        os.path.join(RAW_DATA_path, 'products.csv')
    )

    orders = pd.read_csv(
        os.path.join(RAW_DATA_path, 'orders.csv')
    )

    order_items = pd.read_csv(
        os.path.join(RAW_DATA_path, 'order_items.csv')
    )

    support_tickets = pd.read_csv(
        os.path.join(RAW_DATA_path, 'support_tickets.csv')
    )

    print("All files loaded successfully!")
except NameError:
    print("Error: RAW_DATA_path is not defined. Please ensure the cell defining PROJECT_FOLDER and RAW_DATA_path has been run.")
except FileNotFoundError as e:
    print(f"Error loading CSV: {e}. Please ensure the CSV files exist in '{RAW_DATA_path}'.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
